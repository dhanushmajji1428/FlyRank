## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

 ML Task Type: Classification

My lane is Content Refresh Recommendation.

This is a binary classification problem because the goal is to predict whether a webpage needs a content refresh or does not need a content refresh based on its historical performance metrics.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

The target variable is whether a webpage needs a content refresh.

Since the dataset does not contain a direct label, a proxy target can be created using historical performance metrics such as clicks, impressions, CTR, sessions, and average search position. Pages with declining performance can be labeled as Needs Refresh (1), while the remaining pages can be labeled as No Refresh Needed (0).

## 3. Success metric

*One metric you can defend. What number means 'good'?*

The primary evaluation metric is the F1-score because both false positives and false negatives are important.

Additional metrics include:

- Accuracy
- Precision
- Recall
- ROC-AUC

These metrics help evaluate how well the model identifies webpages that truly require content updates.

 4. The Unit of Analysis

Each row in the dataset represents one webpage (content page). Every webpage contains SEO and engagement metrics such as search volume, clicks, impressions, CTR, average position, engagement rate, and trend direction. These features can be used to determine whether the webpage should be refreshed.

```python
# Create a simple proxy target

df["needs_refresh"] = (
    (df["trend_direction"] == "down") &
    (df["avg_position"] > 20)
).astype(int)

df[["trend_direction", "avg_position", "needs_refresh"]].head()
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

5. Why ML Beats a Fixed Rule Here

A fixed rule such as "refresh every page with an average position greater than 20" ignores many important factors that influence content performance. Machine learning can analyze multiple features simultaneously, including search volume, CTR, engagement rate, impressions, average position, and trend direction. This allows the model to identify complex patterns and generate more accurate recommendations than a single manually defined rule.
